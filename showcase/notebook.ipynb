# How fast does a seedling grow?

A tiny model: more hours of sun, faster growth, up to a limit.

```python
import numpy as np
import matplotlib.pyplot as plt

days = np.arange(0, 22)

def height(day, sun_hours):
    rate = 0.9 * (1 - np.exp(-sun_hours / 3))  # cm per day
    return 1.5 + rate * day
```

```python
#| label: fig-growth
#| fig-cap: Seedling height over three weeks at three light levels
fig, ax = plt.subplots(figsize=(6.4, 3.4))
for hours, colour in [(2, '#8b98a5'), (5, '#5fb37c'), (8, '#2f8a52')]:
    ax.plot(days, height(days, hours), label=f'{hours} h of sun', color=colour, linewidth=2.5)
ax.set_xlabel('Days')
ax.set_ylabel('Height (cm)')
ax.legend(frameon=False)
ax.spines[['top', 'right']].set_visible(False)
plt.show()
```
